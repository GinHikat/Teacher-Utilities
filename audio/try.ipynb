```python
import os
import subprocess
import math

def split_mp3_by_duration(input_file, chunk_minutes=45):
    chunk_seconds = chunk_minutes * 60

    # Get total duration
    result = subprocess.run([
        "ffprobe", "-v", "error",
        "-show_entries", "format=duration",
        "-of", "default=noprint_wrappers=1:nokey=1",
        input_file
    ], capture_output=True, text=True)

    total_duration = float(result.stdout.strip())
    num_chunks = math.ceil(total_duration / chunk_seconds)

    print(f"Total duration: {total_duration/60:.1f} min → {num_chunks} chunks of {chunk_minutes} min")

    for i in range(num_chunks):
        start = i * chunk_seconds
        output_file = f"output_part_{i+1}.mp3"

        subprocess.run([
            "ffmpeg", "-i", input_file,
            "-ss", str(start),
            "-t", str(chunk_seconds),
            "-c", "copy",
            "-y", output_file
        ])

        size_mb = os.path.getsize(output_file) / (1024 * 1024)
        print(f"  {output_file}: {size_mb:.1f} MB")
```

```python
split_mp3_by_duration("sound.mp3")
```
